```python
import numpy as np
import matplotlib.pyplot as plt

# Estimated curve: Y_hat = 1 + X - 2(X-1)^2 * I(X >= 1)
def f(X):
    return 1 + X - 2 * (X - 1)**2 * (X >= 1)

# Generate X values
X = np.linspace(-2, 2, 600)
Y = f(X)

fig, ax = plt.subplots(figsize=(11, 7))

# Split into the two segments for coloring
mask_left = X < 1
mask_right = X >= 1

ax.plot(X[mask_left], Y[mask_left], 'b-', linewidth=2.5,
        label=r'$X<1$:  $\hat{Y} = 1 + X$   (line, slope 1, intercept 1)')
ax.plot(X[mask_right], Y[mask_right], 'r-', linewidth=2.5,
        label=r'$X\geq 1$:  $\hat{Y} = -2X^2 + 5X - 1$   (concave-down parabola)')

# Reference axes
ax.axhline(0, color='gray', linestyle='--', linewidth=0.8, alpha=0.6)
ax.axvline(0, color='gray', linestyle='--', linewidth=0.8, alpha=0.6)
ax.axvline(1, color='green', linestyle=':', linewidth=1.0, alpha=0.6)

# Key points

# Intercept (0, 1)
ax.plot(0, 1, 'ko', ms=9, zorder=5)
ax.annotate('Intercept\n$(0,1)$', xy=(0, 1), xytext=(-1.1, 0.15),
            fontsize=11, arrowprops=dict(arrowstyle='->', color='black'))

# Start (-2, -1)
ax.plot(-2, f(-2), 'ko', ms=7)
ax.annotate(f'$(-2,{f(-2):.0f})$', xy=(-2, f(-2)), xytext=(-2.0, -0.9),
            fontsize=11)

# Junction (1, 2)
ax.plot(1, 2, 'go', ms=11, zorder=6)
ax.annotate('Junction $(1,2)$\ncontinuous & smooth\n(slope = 1 on both sides)',
            xy=(1, 2), xytext=(1.15, 0.4),
            fontsize=10, color='green',
            arrowprops=dict(arrowstyle='->', color='green'))

# Peak of the parabola (5/4, 17/8)
peak_x, peak_y = 1.25, 17/8
ax.plot(peak_x, peak_y, 'mo', ms=9, zorder=6)
ax.annotate(f'Peak $(5/4,\\, 17/8)$\nslope = 0',
            xy=(peak_x, peak_y), xytext=(0.35, 2.55),
            fontsize=11, color='magenta',
            arrowprops=dict(arrowstyle='->', color='magenta'))

# Endpoint (2, 1)
ax.plot(2, f(2), 'co', ms=8)
ax.annotate(f'$(2,{f(2):.0f})$', xy=(2, f(2)), xytext=(1.9, 0.55),
            fontsize=11, color='teal')

# Slope annotation on the linear piece
ax.annotate('slope = 1', xy=(-0.6, 0.4), fontsize=12, color='blue',
            rotation=40, ha='center')

# Labels / title
ax.set_xlabel('$X$', fontsize=13)
ax.set_ylabel(r'Estimated $\hat{Y}$', fontsize=13)
ax.set_title(
    r'Estimated curve:  $\hat{Y} = 1 + X - 2(X-1)^2\,\mathbb{I}(X\geq 1)$' + '\n'
    r'with $b_1(X)=X,\quad b_2(X)=(X-1)^2\,\mathbb{I}(X\geq 1)$',
    fontsize=13
)
ax.legend(loc='lower right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim([-2.2, 2.2])
ax.set_ylim([-1.5, 2.85])

plt.tight_layout()
plt.show()
```
